# Using python speech recognition library

## Reading an audio file

```python
# Import the SpeechRecognition library
import speech_recognition as sr
recognizer = sr.Recognizer()

# The loudness of audio which is consider speech
# below = silent
# above = speech
recognizer.energy_threshold = 300 # 0 to 4000

# Read an audio file
recorded_sound = sr.AudioFile("output.wav")
type(recorded_sound)
```

```python
# Try speech recognition
text = recognizer.recognize_google(audio_data=recorded_sound, language="en-US")
print(text)
```

## Sound to text

```python
# We need an audio data type
with recorded_sound as source:
    recorded_sound_audio = recognizer.record(source, duration=None, offset=None)
type(recorded_sound_audio)
```

```python
# Try speech recognizition
text = recognizer.recognize_google(audio_data=recorded_sound_audio, language="en-US")
print(text)
```

## Modify the start or end

```python
# We need an audio data type
with recorded_sound as source:
    recorded_sound_audio =recognizer.record(source, duration=3, offset=None)
type(recorded_sound_audio)
```

```python
# Try speech recognizition
text = recognizer.recognize_google(audio_data=recorded_sound_audio, language="en-US")
print(text)
```

## Dealing with different kinds of audio
- What language? specify in language="ja" or "en-US"
- Non speech audio? comes back empty
- Noisy audio? adjust for ambient noise

```python
# Show all the potential transcriptions
text = recognizer.recognize_google(audio_data=recorded_sound_audio, language="en-US", show_all=True)
print(text)
```

```python
# Multiple speakers
# google api returns 1 blob of text for all speakers
# alternative record separetely for each speaker
```

```python
# Noisy audio
noisy_sound = sr.AudioFile("output.wav")
with noisy_sound as source:
    recognizer.adjust_for_ambient_noise(source, duration=0.5) # 0.5 to 1sec
    noisy_sound_audio = recognizer.record(source)
recognizer.recognize_google(noisy_sound_audio)
```
